```python
import gc, torch
from vllm import LLM, SamplingParams
import time
import pandas as pd
import numpy as np
```

```python


# Clear cache and collect garbage to get accurate VRAM measurements
gc.collect()
torch.cuda.empty_cache()

print(f"GPU: {torch.cuda.get_device_name(0)} | VRAM: {torch.cuda.get_device_properties(0).total_memory / 1e9:.1f} GB")

# Define common parameters for the experiment
main_model_name = "Qwen/Qwen2.5-1.5B-Instruct"
prompts = ["Объясни кратко, что такое спекулятивный декодинг."]
sampling_params = SamplingParams(max_tokens=50, temperature=0, logprobs=None)

# Function to run an experiment and collect metrics
def run_experiment(
    main_model: str,
    draft_model: str = None,
    num_speculative_tokens: int = 5,
    method: str = "draft_model",
    max_model_len: int = 256,
    gpu_memory_utilization: float = 0.85,
    enforce_eager: bool = True,
    dtype: str = "float16",
    prompts: list = None,
    sampling_params: SamplingParams = None,
    experiment_name: str = "Experiment",
    num_runs_per_config: int = 10 # Added parameter for number of iterations per config
):
    print(f"\nStarting {num_runs_per_config} iterations for: {experiment_name}")

    all_iteration_results = []


    # Reset peak memory stats before loading models for accurate measurement for this iteration
    torch.cuda.reset_peak_memory_stats()

    llm_config = {
        "model": main_model,
        "max_model_len": max_model_len,
        "gpu_memory_utilization": gpu_memory_utilization,
        "enforce_eager": enforce_eager,
        "dtype": dtype,
        "tensor_parallel_size": 1,
        "disable_log_stats": False # Keep logging for metrics collection
    }

    if draft_model:
        llm_config["speculative_config"] = {
            "model": draft_model,
            "num_speculative_tokens": num_speculative_tokens,
            "method": method
        }
        print(f"Main Model: {main_model}, Draft Model: {draft_model} (dtype={dtype})")
    else:
        print(f"Main Model: {main_model}, Speculative Decoding: DISABLED (Baseline), (dtype={dtype})")

    # Clear cache and collect garbage before LLM init for cleaner VRAM start
    gc.collect()
    torch.cuda.empty_cache()

    llm = LLM(**llm_config)

    # Warmup для выделения всей памяти
    _ = llm.generate(prompts[0], SamplingParams(max_tokens=10))

        # ... после warmup
    torch.cuda.synchronize()

    # Используем reserved, так как vLLM резервирует пулы памяти заранее
    vram_reserved_gb = torch.cuda.memory_reserved() / (1024**3)

    # Альтернатива: через nvidia-smi (самый точный способ для vLLM)
    import subprocess
    result = subprocess.run(['nvidia-smi', '--query-gpu=memory.used', '--format=csv,nounits,noheader'], stdout=subprocess.PIPE)
    vram_used_mb = float(result.stdout.decode('utf-8').strip())
    vram_smi_gb = vram_used_mb / 1024

    print(f"✅ VRAM Reserved (PyTorch): {vram_reserved_gb:.2f} GB")
    print(f"✅ VRAM Used (nvidia-smi): {vram_smi_gb:.2f} GB")

    vram_allocated_gb = torch.cuda.memory_allocated() / (1024**3)
    vram_reserved_gb = torch.cuda.memory_reserved() / (1024**3)

    print(f"✅ Models loaded. VRAM Allocated: {vram_allocated_gb:.2f} GB")

    # Measure VRAM after LLM init (this captures memory used by loaded models)
    vram_after_init_gb = torch.cuda.max_memory_allocated() / (1024**3)

    print(f"✅ Models loaded. Peak VRAM after init: {vram_after_init_gb:.2f} GB. Starting generation...")


    for i in range(num_runs_per_config):
        print(f"  Iteration {i+1}/{num_runs_per_config}")
        current_experiment_name = f"{experiment_name} (Iteration {i+1})"

        start_time = time.time()
        outputs = llm.generate(prompts, sampling_params)
        end_time = time.time()
        elapsed_time = end_time - start_time

        # Collect metrics from vLLM engine
        metrics = llm.llm_engine.get_metrics()
        metrics_dict = {m.name: m for m in metrics if 'vllm:' in m.name}

        num_generated_tokens = sum(len(output.outputs[0].token_ids) for output in outputs)
        latency_ms_per_token = (elapsed_time * 1000 / num_generated_tokens) if num_generated_tokens > 0 else 0.0

        num_drafts_metric = metrics_dict.get('vllm:spec_decode_num_drafts')
        num_drafts = num_drafts_metric.value if num_drafts_metric else 0

        num_draft_tokens_metric = metrics_dict.get('vllm:spec_decode_num_draft_tokens')
        num_draft_tokens = num_draft_tokens_metric.value if num_draft_tokens_metric else 0

        num_accepted_tokens_metric = metrics_dict.get('vllm:spec_decode_num_accepted_tokens')
        num_accepted_tokens = num_accepted_tokens_metric.value if num_accepted_tokens_metric else 0

        # Acceptance length: Average accepted tokens per draft step
        avg_acceptance_length = (num_accepted_tokens / num_drafts) if num_drafts > 0 else 0.0

        # Draft Generation Speed: Approximation based on total draft tokens and total generation time
        draft_gen_speed_tokens_per_sec = (num_draft_tokens / elapsed_time) if elapsed_time > 0 else 0.0

        iteration_result = {
            "Experiment": current_experiment_name,
            "Main Model": main_model,
            "Draft Model": draft_model if draft_model else "N/A (Baseline)",
            "Total Time (s)": elapsed_time,
            "Generated Tokens": num_generated_tokens,
            "Latency (ms/token)": latency_ms_per_token,
            "Num Drafts": num_drafts,
            "Num Draft Tokens": num_draft_tokens,
            "Num Accepted Tokens": num_accepted_tokens,
            "Avg Acceptance Length": avg_acceptance_length,
            "Draft Gen Speed (tokens/s)": draft_gen_speed_tokens_per_sec,
            "VRAM Peak (GB)": vram_after_init_gb, # Peak VRAM observed after model loading
            "VRAM Allocated (GB)": vram_allocated_gb,
            "VRAM Reserved (GB)": vram_reserved_gb,
            "VRAM smi (GB)": vram_smi_gb,
        }
        all_iteration_results.append(iteration_result)

    # Clear LLM to free memory for the next run
    del llm
    gc.collect()
    torch.cuda.empty_cache()

    # Calculate average metrics
    if not all_iteration_results:
        return {} # Should not happen if num_runs_per_config > 0

    avg_results = {}
    # Identify numeric keys to average
    numeric_keys = [key for key, value in all_iteration_results[0].items() if isinstance(value, (int, float))]


    # Keep non-numeric fields from the first iteration
    for key in all_iteration_results[0]:
        if not isinstance(all_iteration_results[0][key], (int, float)):
            avg_results[key] = all_iteration_results[0][key]

    for key in numeric_keys:
        avg_results[key] = np.mean([res[key] for res in all_iteration_results])

    # Modify the Experiment name to reflect it's an average
    avg_results["Experiment"] = f"Average {experiment_name} (over {num_runs_per_config} iterations)"

    return avg_results
```

```python


# Experiment configurations
# Note: vLLM's `speculative_config` does not directly support applying different
# quantization types (e.g., 'awq', 'gptq') to the draft model alone.
# The 'dtype' parameter in the main LLM constructor applies to both models.
# To experiment with different quantizations for the draft model, one would
# typically need to use pre-quantized versions of the draft model (if available
# on Hugging Face) or define the draft model as a separate LLM with its own
# quantization, which is not how `speculative_config` is designed.
experiment_configs = [
    {
        "name": "Baseline (No Speculative Decoding)",
        "draft_model": None,
        "dtype": "float16" # Main model dtype
    },
    {
        "name": "Speculative Decoding",
        "draft_model": "Qwen/Qwen2.5-0.5B-Instruct",
        "num_speculative_tokens": 8,
        "dtype": "float16"
    }
]

results = []
# Assuming num_iterations is defined globally, as seen in kernel state.
# If not, it would need to be defined here.
# For example: num_iterations = 10
num_iterations = 30
for config in experiment_configs:
    result = run_experiment(
        main_model=main_model_name,
        draft_model=config.get("draft_model"),
        num_speculative_tokens=config.get("num_speculative_tokens", 3),
        dtype=config.get("dtype", "float16"),
        prompts=prompts,
        sampling_params=sampling_params,
        experiment_name=config["name"],
        num_runs_per_config=num_iterations # Pass the global num_iterations
    )
    results.append(result)

# Display results in a clear markdown table
results_df = pd.DataFrame(results)
# Round numeric columns for cleaner output
numeric_cols = results_df.select_dtypes(include=[np.number]).columns
results_df[numeric_cols] = results_df[numeric_cols].round(2)

print("\n--- Experiment Results ---")
print(results_df.to_markdown(index=False))
```

## Общий эксперимент

```python
import gc, torch
from vllm import LLM, SamplingParams
import time
import pandas as pd
import numpy as np
from datasets import load_dataset
import subprocess


# Clear cache and collect garbage to get accurate VRAM measurements
gc.collect()
torch.cuda.empty_cache()

print(f"GPU: {torch.cuda.get_device_name(0)} | VRAM: {torch.cuda.get_device_properties(0).total_memory / 1e9:.1f} GB")

# Load the dataset from Hugging Face
dataset = load_dataset("zzzhr97/SpecBench")

prompts_by_category = {}
for split_name in dataset:
    category_prompts = []
    for item in dataset[split_name]:
        category_prompts.append(item['prompt'])
    prompts_by_category[split_name] = category_prompts

print(f"Loaded {sum(len(v) for v in prompts_by_category.values())} prompts across {len(prompts_by_category)} categories.")

# Define common parameters for the experiment
main_model_name = "Qwen/Qwen2.5-1.5B-Instruct"
sampling_params = SamplingParams(max_tokens=50, temperature=0.7, top_p=0.9)

# Global parameters for LLM initialization that are constant across all configs
global_max_model_len = 350
global_gpu_memory_utilization = 0.7
global_enforce_eager = True
global_dtype = "float16"

def run_experiment(
    llm_instance: LLM, # Pass the initialized LLM instance
    config_details: dict, # Pass the config dictionary for model details and logging
    vram_smi_gb: float, # Pass VRAM measured after model load
    prompts_list: list = None,
    sampling_params: SamplingParams = None,
    experiment_name: str = "Experiment", # This will now include category name
    num_runs_per_config: int = 1, # Still 1 in current setup
    category_name_for_results: str = "" # For adding category to results
):
    print(f"Starting experiment: {experiment_name}")

    # Extract model-specific details from config_details for logging and prompt truncation
    main_model_name_log = config_details["main_model_name_for_logging"]
    draft_model_log = config_details.get("draft_model")
    num_speculative_tokens_log = config_details.get("num_speculative_tokens", 0 if not draft_model_log else 5)
    max_model_len_for_prompts = config_details.get("max_model_len_for_prompts", global_max_model_len)

    # Apply max_model_len to prompts to avoid issues with long prompts
    prompts_list = [prompt[:max_model_len_for_prompts-50] for prompt in prompts_list]

    all_prompt_results = [] # Collects results for each prompt across all iterations

    # Use the passed LLM instance
    llm = llm_instance

    for i in range(num_runs_per_config):
        print(f"  Iteration {i+1}/{num_runs_per_config}")

        start_time_batch = time.time()
        outputs = llm.generate(prompts_list, sampling_params, use_tqdm=True)
        end_time_batch = time.time()
        total_batch_time = end_time_batch - start_time_batch

        metrics_vllm_engine = llm.llm_engine.get_metrics()
        metrics_dict_vllm = {m.name: m for m in metrics_vllm_engine if 'vllm:' in m.name}

        num_drafts = metrics_dict_vllm.get('vllm:spec_decode_num_drafts').value if metrics_dict_vllm.get('vllm:spec_decode_num_drafts') else 0
        num_draft_tokens = metrics_dict_vllm.get('vllm:spec_decode_num_draft_tokens').value if metrics_dict_vllm.get('vllm:spec_decode_num_draft_tokens') else 0
        num_accepted_tokens_total = metrics_dict_vllm.get('vllm:spec_decode_num_accepted_tokens').value if metrics_dict_vllm.get('vllm:spec_decode_num_accepted_tokens') else 0

        avg_acceptance_length_batch = (num_accepted_tokens_total / num_drafts) if num_drafts > 0 else 0.0
        draft_gen_speed_tokens_per_sec_batch = (num_draft_tokens / total_batch_time) if total_batch_time > 0 else 0.0
        acceptance_rate_batch = (num_accepted_tokens_total / num_draft_tokens) * 100 if num_draft_tokens > 0 else 0.0

        for idx, output in enumerate(outputs):
            prompt_text = output.prompt
            generated_text = output.outputs[0].text
            num_generated_tokens_per_prompt = len(output.outputs[0].token_ids)

            request_first_token_time = output.metrics.first_token_ts
            request_last_token_time = output.metrics.last_token_ts

            total_time_s_per_prompt = request_last_token_time - request_first_token_time if request_first_token_time and request_last_token_time else 0.0
            latency_ms_per_token_per_prompt = (total_time_s_per_prompt * 1000 / num_generated_tokens_per_prompt) if num_generated_tokens_per_prompt > 0 else 0.0
            main_model_gen_speed_tokens_per_sec_per_prompt = (num_generated_tokens_per_prompt / total_time_s_per_prompt) if total_time_s_per_prompt > 0 else 0.0

            prompt_result = {
                "Experiment": f"{experiment_name} (Iteration {i+1})",
                "Prompt Index": idx,
                "Prompt": prompt_text,
                "Main Model": main_model_name_log,
                "Draft Model": draft_model_log if draft_model_log else "N/A (Baseline)",
                "Num Speculative Tokens": num_speculative_tokens_log,
                "VRAM smi (GB)": vram_smi_gb, # Constant for all prompts in this model load
                "Total Time (s)": total_time_s_per_prompt,
                "Latency (ms/token)": latency_ms_per_token_per_prompt,
                "Num Drafts": num_drafts, # Aggregate for the batch
                "Num Draft Tokens": num_draft_tokens, # Aggregate for the batch
                "Num Accepted Tokens": num_accepted_tokens_total, # Aggregate for the batch
                "Avg Acceptance Length": avg_acceptance_length_batch, # Aggregate for the batch
                "Acceptance rate": acceptance_rate_batch, # Aggregate for the batch
                "Draft Gen Speed (tokens/s)": draft_gen_speed_tokens_per_sec_batch, # Aggregate for the batch
                "Main Model Gen Speed (tokens/s)": main_model_gen_speed_tokens_per_sec_per_prompt,
                "Generated Output": generated_text, # For debugging/inspection
                "Category": category_name_for_results # Add category info
            }
            all_prompt_results.append(prompt_result)

    return all_prompt_results

# Experiment configurations
experiment_configs = []

# Config 1: Baseline (No Speculative Decoding)
experiment_configs.append(
    {
        "name": "Baseline (No Speculative Decoding)",
        "draft_model": None,
        "num_speculative_tokens": 0 # Not applicable
    }
)

# Config 2: Base + Int8 Draft Model with varying num_speculative_tokens
for n_spec_tokens in [2, 3, 4, 5]:
    experiment_configs.append(
        {
            "name": f"Speculative Decoding (Int8, {n_spec_tokens} tokens)",
            "draft_model": "Qwen/Qwen2.5-0.5B-Instruct-GPTQ-Int8",
            "num_speculative_tokens": n_spec_tokens
        }
    )

# Config 3: Base + Int4 Draft Model with varying num_speculative_tokens
for n_spec_tokens in [2, 3, 4, 5]:
    experiment_configs.append(
        {
            "name": f"Speculative Decoding (Int4, {n_spec_tokens} tokens)",
            "draft_model": "Qwen/Qwen2.5-0.5B-Instruct-GPTQ-Int4",
            "num_speculative_tokens": n_spec_tokens
        }
    )

all_results_df_list = []
num_runs_per_config_iteration = 1 # Each configuration will be run once over the full dataset of prompts

# Outer loop: Iterate through each *model configuration* (Main + Draft if applicable)
for config in experiment_configs:
    print(f"\n--- Loading Model for: {config['name']} ---")

    # Clear cache and collect garbage before LLM init for cleaner VRAM start for this specific model config
    gc.collect()
    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats()

    llm_config = {
        "model": main_model_name,
        "max_model_len": global_max_model_len,
        "gpu_memory_utilization": global_gpu_memory_utilization,
        "enforce_eager": global_enforce_eager,
        "dtype": global_dtype,
        "tensor_parallel_size": 1,
        "disable_log_stats": False, # Keep logging for metrics collection
        "enable_prefix_caching": False,
        "max_num_seqs": 64, # Огранииваем количество одновременных запросов (размер батча)
    }

    current_draft_model = config.get("draft_model")
    if current_draft_model:
        llm_config["speculative_config"] = {
            "model": current_draft_model,
            "num_speculative_tokens": config.get("num_speculative_tokens"),
            "method": "draft_model"
        }
        print(f"Main Model: {main_model_name}, Draft Model: {current_draft_model} (dtype={llm_config['dtype']}, num_spec_tokens={config.get('num_speculative_tokens')})")
    else:
        print(f"Main Model: {main_model_name}, Speculative Decoding: DISABLED (Baseline), (dtype={llm_config['dtype']})")

    # Initialize the LLM once for this model configuration
    llm = LLM(**llm_config)

    # Warmup for memory allocation and VRAM measurement after loading each distinct model configuration
    if prompts_by_category:
        first_category_prompts_list = next(iter(prompts_by_category.values()))
        if first_category_prompts_list:
            # Use first prompt of first category for warmup, truncate to global_max_model_len
            warmup_prompt = first_category_prompts_list[0][:global_max_model_len-50]
            _ = llm.generate([warmup_prompt], SamplingParams(max_tokens=10))
    torch.cuda.synchronize()

    result = subprocess.run(['nvidia-smi', '--query-gpu=memory.used', '--format=csv,nounits,noheader'], stdout=subprocess.PIPE)
    vram_used_mb = float(result.stdout.decode('utf-8').strip())
    vram_smi_gb = vram_used_mb / 1024
    print(f"✅ VRAM Used (nvidia-smi) for this config: {vram_smi_gb:.2f} GB")


    # Inner loop: Iterate through each category, using the *same loaded LLM* instance
    for category_name, prompts_for_category in prompts_by_category.items():
        print(f"\n====== Running: {config['name']} for Category: {category_name} ======")

        # Prepare config_details to pass to run_experiment for logging
        current_config_details_for_logging = {
            "main_model_name_for_logging": main_model_name,
            "draft_model": config.get("draft_model"),
            "num_speculative_tokens": config.get("num_speculative_tokens", 0 if not current_draft_model else 5),
            "max_model_len_for_prompts": global_max_model_len
        }

        results_for_category_config = run_experiment(
            llm_instance=llm, # Pass the initialized LLM
            config_details=current_config_details_for_logging, # Pass details for logging
            vram_smi_gb=vram_smi_gb, # Pass the VRAM measured after loading this model
            prompts_list=prompts_for_category,
            sampling_params=sampling_params,
            experiment_name=f"{config['name']} - {category_name}",
            num_runs_per_config=num_runs_per_config_iteration,
            category_name_for_results=category_name
        )
        all_results_df_list.extend(results_for_category_config)

    # Clear LLM to free memory for the next model configuration to be loaded
    del llm
    gc.collect()
    torch.cuda.empty_cache()

# Display results in a clear markdown table
final_results_df = pd.DataFrame(all_results_df_list)

# Round numeric columns for cleaner output
numeric_cols = final_results_df.select_dtypes(include=[np.number]).columns
final_results_df[numeric_cols] = final_results_df[numeric_cols].round(2)

# print("\n--- Final Experiment Results (Per Prompt) ---")
# print(final_results_df.to_markdown(index=False))
```
